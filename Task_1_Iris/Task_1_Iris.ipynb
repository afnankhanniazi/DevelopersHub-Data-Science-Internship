# Task 1: Exploring and Visualizing a Simple Dataset
**Objective:** To understand how to read, summarize, and visualize a dataset using pandas, matplotlib, and seaborn.
**Dataset:** The Iris Dataset, which contains measurements of sepal and petal dimensions for three different species of iris flowers.

```python
# Import the required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Iris dataset built into seaborn
df = sns.load_dataset('iris')

# 1. Display the dataset structure (rows and columns)
print("Dataset Shape:", df.shape)

# 2. Display the column names
print("\nDataset Columns:", df.columns.tolist())

# 3. Display the first 5 rows
print("\nFirst 5 Rows:")
display(df.head())
```

```python
# Set up a clean grid style for our charts
sns.set_theme(style="whitegrid")

# Create a figure with 3 subplots side-by-side
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# 1. Scatter Plot: Analyzing relationships between sepal length and width
sns.scatterplot(ax=axes[0], data=df, x='sepal_length', y='sepal_width', hue='species')
axes[0].set_title('Scatter Plot: Sepal Length vs. Width')

# 2. Histogram: Examining the distribution of petal lengths
sns.histplot(ax=axes[1], data=df, x='petal_length', kde=True, hue='species')
axes[1].set_title('Histogram: Petal Length Distribution')

# 3. Box Plot: Detecting outliers and spread of petal widths by species
sns.boxplot(ax=axes[2], data=df, x='species', y='petal_width')
axes[2].set_title('Box Plot: Petal Width Spread & Outliers')

# Display the charts neatly
plt.tight_layout()
plt.show()
```

```python
# 1. The Pairplot: A comprehensive grid of all relationships
print("Generating Advanced Pairplot...")
sns.pairplot(df, hue='species', palette='husl', markers=["o", "s", "D"])
plt.suptitle("Advanced EDA: Pairwise Relationships Across All Features", y=1.02)
plt.show()

# 2. The Correlation Heatmap: Mathematically proving the relationships
print("\nGenerating Correlation Heatmap...")
plt.figure(figsize=(8, 6))

# We must drop the text column ('species') because correlation only works on numbers
numeric_df = df.drop('species', axis=1)

# Create the heatmap
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', linewidths=2, fmt=".2f")
plt.title("Advanced EDA: Feature Correlation Heatmap")
plt.show()
```

### Conclusion & Advanced Insights

Through our foundational Exploratory Data Analysis, we established clear visual distinctions between the three iris species, most notably that the `setosa` species is distinctly isolated from the others based on its smaller petal and sepal dimensions.

By extending our analysis with advanced visual techniques, we uncovered deeper mathematical relationships within the dataset:
* **Multidimensional Clustering:** The Pairplot confirmed that while `versicolor` and `virginica` share some dimensional overlap, they can still be effectively clustered when evaluating multiple features simultaneously.
* **Feature Correlation:** The Correlation Heatmap mathematically proved a nearly perfect positive correlation (0.96) between petal length and petal width, indicating that as these flowers grow longer petals, they predictably grow wider petals.

This multi-layered approach not only highlights the dataset's underlying structure but also demonstrates a robust, industry-standard methodology for feature evaluation prior to applying any predictive machine learning algorithms.
